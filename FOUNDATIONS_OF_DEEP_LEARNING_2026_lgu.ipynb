<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">
Two keywords do all the work: <code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">class</code> defines the blueprint,
and <code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">def</code> gives it a behaviour called a <b>method</b>.
</div>

```python
# STEP 1 — define the blueprint
class Dog:
    def bark(self):        # bark = a method (function inside a class)
        print('woof')
```

```python
# STEP 2 — build an object, then make it bark
german_shephard = Dog()
german_shephard.bark()     # woof
```

```python
# STEP 3 — same blueprint makes another object
a_shephard = Dog()
a_shephard.bark()          # woof
```

<div style="background:#FFFFFF;border-left:8px solid #FFB000;border-radius:8px;padding:14px 22px;font-family:Segoe UI,system-ui,sans-serif;font-size:21px;color:#0B1E3D;font-weight:600">
  🔑 <code style="background:#e6f0f6;color:#0a3d62;padding:2px 8px;border-radius:5px">self</code> = "this object's own". It is the first input of every method.
</div>

<div style="background:#0a3d62;border-radius:10px;padding:14px 24px;font-family:Segoe UI,system-ui,sans-serif;margin-top:10px">
  <span style="color:#FFB000;font-size:30px;font-weight:900">02</span>
  <span style="color:#FFFFFF;font-size:27px;font-weight:800;margin-left:12px">__init__ — give the object a memory</span>
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">__init__</code> is the <b>constructor</b> — it runs the moment an object is created and stores its starting data.
</div>

```python
# STEP 1 — a Dog that remembers its name and temperature
class Dog:
    def __init__(self, petname, temp):    # constructor
        self.name = petname               # attribute (data)
        self.temprature = temp

    def status(self):                     # getter — show the data
        print('petname is', self.name)
        print('pet temprature is', self.temprature)

    def settemprature(self, temp):        # setter — change the data
        self.temprature = temp
```

```python
# STEP 2 — one blueprint, three different objects
german_shephard = Dog('German shephard', 35)
lessi           = Dog('lessi', 39)
a_shephard      = Dog('a', 40)
```

```python
# STEP 3 — each object remembers its own data
german_shephard.status()
lessi.status()
a_shephard.status()
```

```python
# STEP 1 — define the Bag blueprint
class Bag:
    def __init__(self, color, price, weight):
        self.color  = color
        self.price  = price
        self.weight = weight

    def status(self):
        print('The bag is of', self.color)
        print('the bag costs', self.price)
        print('the bag has weight', self.weight)

    def setcolor(self, color):
        self.color = color
```

```python
# STEP 2 — build the object and show it
nike = Bag('blue color', 2000, 40)
nike.status()
```

```python
# STEP 3 — change the colour, show again
nike.setcolor('red')
nike.status()
```

```python
# STEP 1 — a Student blueprint with real fields
class Student:
    def __init__(self, name, registration_no, subjects, CGPA, grade):
        self.name = name
        self.registration_no = registration_no
        self.subjects = subjects
        self.CGPA  = CGPA
        self.grade = grade

    def status(self):
        print('My name is:', self.name)
        print('My registration no is:', self.registration_no)
        print('I have chosen the subjects:', self.subjects)
        print('My CGPA is:', self.CGPA)
        print('I got the grade:', self.grade)

    def setname(self, name):
        self.name = name
    def setregistration_no(self, registration_no):
        self.registration_no = registration_no
```

```python
# STEP 2 — build a real student object
student = Student('Momina Arshad', 10,
                  ['Functional Analysis', 'Advanced Differential Equations',
                   'Partial Differential Equations', 'Advanced Linear Algebra'],
                  3.57, 'B+')
student.status()
```

```python
# STEP 3 — update name and registration no
student.setname('Amna')
student.setregistration_no(20)
student.status()
```

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">
The real power of classes: make <b>many objects</b>, store them in a <b>list</b>, and process them with a <b>loop</b>.
</div>

```python
# STEP 1 — a small class with a decision method
class Learner:
    def __init__(self, name, cgpa):
        self.name = name
        self.cgpa = cgpa

    def grade(self):                 # logic: CGPA -> grade
        if   self.cgpa >= 3.5: return 'A'
        elif self.cgpa >= 3.0: return 'B'
        elif self.cgpa >= 2.0: return 'C'
        else:                  return 'F' 
```

```python
# STEP 2 — a list of objects, processed by a loop
batch = [
    Learner('Momina', 3.57),
    Learner('Amna',   3.10),
    Learner('Ahmad',  2.40),
    Learner('Ali',    1.80),
]

for L in batch:
    print(f"{L.name:8} CGPA={L.cgpa}  ->  Grade {L.grade()}")
```

```python
# STEP 3 — logic: average CGPA of only A/B students
good = [L.cgpa for L in batch if L.grade() in ('A', 'B')]
print("Top students :", len(good))
print("Average CGPA :", round(sum(good) / len(good), 2))
```

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">
Everything we learned about <b>classes</b> now builds a real <b>neural network</b>. It is just a bigger class:
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">__init__</code> sets up the weights,
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">train</code> teaches it,
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">query</code> asks it, and
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">backquery</code> shows what it imagined.
Run the import cell first.
</div>

```python
import numpy as np
import scipy.special

# compatibility: np.asfarray was removed in NumPy 2.0 — keep old notebooks working
if not hasattr(np, "asfarray"):
    np.asfarray = lambda a, dtype=float: np.asarray(a, dtype=dtype)
```

```python
class Neuralnetwork:
    def __init__(self, inodes, hnodes, onodes, learningrate):
        self.inputnodes=inodes
        self.hiddennodes=hnodes
        self.outputnodes=onodes
        self.lr=learningrate

        self.wih=np.random.normal(0.0, pow(self.hiddennodes, -0.5), (self.hiddennodes, self.inputnodes))
        self.who=np.random.normal(0.0, pow(self.outputnodes, -0.5), (self.outputnodes, self.hiddennodes))

        self.activation=lambda x: scipy.special.expit(x)
        self.inverse_activation_function = lambda x: scipy.special.logit(x)

        pass

    def train(self, model_input, labeled_output):

        inputs=np.array(model_input, ndmin=2).T
        lebeledouts=np.array(labeled_output, ndmin=2).T

        hidden_input=np.dot(self.wih, inputs)
        hidden_output=self.activation(hidden_input)

        final_input=np.dot(self.who, hidden_output)
        final_output= self.activation(final_input)

        output_error=lebeledouts-final_output
        hidden_error=np.dot(self.who.T,output_error)

        self.who+=self.lr*np.dot(output_error*final_output*(1-final_output), np.transpose(hidden_output))
        self.wih+=self.lr*np.dot(hidden_error*hidden_output*(1-hidden_output), np.transpose(inputs))
        pass

    def query(self,model_input):
        inputs=np.array(model_input, ndmin=2).T

        hidden_input=np.dot(self.wih, inputs)
        hidden_output=self.activation(hidden_input)

        final_input=np.dot(self.who, hidden_output)
        final_output= self.activation(final_input)
        return final_output

    #backwards query to have an idea what NN has learned
    def backquery(self, targets_list):
        # transpose the targets list to a vertical array
        final_outputs = np.array(targets_list, ndmin=2).T

        # calculate the signal into the final output layer
        final_inputs = self.inverse_activation_function(final_outputs)
        # calculate the signal out of the hidden layer
        hidden_outputs = np.dot(self.who.T, final_inputs)
        # scale them back to 0.01 to .99
        hidden_outputs -= np.min(hidden_outputs)
        hidden_outputs /= np.max(hidden_outputs)
        hidden_outputs *= 0.98
        hidden_outputs += 0.01

        # calculate the signal into the hidden layer
        hidden_inputs = self.inverse_activation_function(hidden_outputs)

        # calculate the signal out of the input layer
        inputs = np.dot(self.wih.T, hidden_inputs)
        # scale them back to 0.01 to .99
        inputs -= np.min(inputs)
        inputs /= np.max(inputs)
        inputs *= 0.98
        inputs += 0.01

        return inputs
```

  💡 Same idea as Dog and Student — a <b>class</b> with <code style="background:#e6f0f6;color:#0a3d62;padding:1px 6px;border-radius:4px">__init__</code> and methods. Just now the &quot;data&quot; are weight matrices, and the methods do the learning.
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:15px;color:#33475a">Finally, just like we made a Dog object, we make one <b>neural network object</b> from the class:</div>

```python
# make one neural network object from the class
# (input nodes, hidden nodes, output nodes, learning rate)
NN = Neuralnetwork(3, 3, 3, 0.3)
print(NN)   # the object is ready
```

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">

The toy network above had 3 inputs. A 28&times;28 picture has <b>784</b> pixels, so the same class
simply grows: <b>784 inputs</b>, <b>two hidden layers</b>, and <b>10 outputs</b> &mdash; one for every digit 0&ndash;9.
We also add <code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">matplotlib</code> so we can see the images.

<div style="background:#0a3d62;border-radius:10px;padding:14px 24px;font-family:Segoe UI,system-ui,sans-serif;margin-top:10px">
  <span style="color:#FFB000;font-size:30px;font-weight:900">10</span>
  <span style="color:#FFFFFF;font-size:27px;font-weight:800;margin-left:12px">The network class &mdash; two hidden layers</span>
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">

Exactly the same shape as <b>Dog</b> and <b>Student</b>:
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">__init__</code> stores the data (here the <b>weight matrices</b>),
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">train</code> changes that data (learning), and
<code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">query</code> shows the answer. The only new thing is <b>one more hidden layer</b>.

</div>

<div style="background:#0B1E3D;border-radius:14px;padding:24px 20px;font-family:Segoe UI,system-ui,sans-serif;text-align:center;color:#FFFFFF">
  <div style="font-size:15px;letter-spacing:3px;color:#00D9E9;font-weight:800;margin-bottom:16px">ARCHITECTURE</div>

  <table style="border-collapse:separate;border-spacing:6px;margin:0 auto">
  <tr>
    <td style="background:#FFB000;color:#0B1E3D;padding:12px 20px;border-radius:10px;font-weight:800;font-size:20px;line-height:1.3">INPUT<br>784</td>
    <td style="color:#00D9E9;font-size:26px;font-weight:800;padding:0 6px">&rarr;</td>
    <td style="background:#00D9E9;color:#0B1E3D;padding:12px 20px;border-radius:10px;font-weight:800;font-size:20px;line-height:1.3">HIDDEN 1<br>500</td>
    <td style="color:#00D9E9;font-size:26px;font-weight:800;padding:0 6px">&rarr;</td>
    <td style="background:#00D9E9;color:#0B1E3D;padding:12px 20px;border-radius:10px;font-weight:800;font-size:20px;line-height:1.3">HIDDEN 2<br>300</td>
    <td style="color:#00D9E9;font-size:26px;font-weight:800;padding:0 6px">&rarr;</td>
    <td style="background:#00E08A;color:#0B1E3D;padding:12px 20px;border-radius:10px;font-weight:800;font-size:20px;line-height:1.3">OUTPUT<br>10</td>
  </tr>
  </table>

  <div style="margin-top:18px;font-size:17px;color:#BcD6E6">
    three weight matrices: <code style="color:#00D9E9">wih1</code> (500&times;784) &nbsp;&middot;&nbsp;
    <code style="color:#00D9E9">wh1h2</code> (300&times;500) &nbsp;&middot;&nbsp;
    <code style="color:#00D9E9">wh2o</code> (10&times;300)
  </div>
</div>

```python
# STEP 1 — the blueprint (class), the same Dog/Student pattern on a bigger scale
class NeuralNetwork:

    def __init__(self, input_nodes, hidden1_nodes, hidden2_nodes, output_nodes, learning_rate):
        self.inodes  = input_nodes
        self.h1nodes = hidden1_nodes
        self.h2nodes = hidden2_nodes
        self.onodes  = output_nodes
        self.lr      = learning_rate

        # np.random.normal(mean, std, size)
        # mean = 0.0
        # std  = 1/sqrt(number of destination nodes)
        # size = (rows, columns)

        # Input -> Hidden Layer 1 weights
        self.wih1  = np.random.normal(0.0, pow(self.h1nodes, -0.5), (self.h1nodes, self.inodes))

        # Hidden Layer 1 -> Hidden Layer 2 weights
        self.wh1h2 = np.random.normal(0.0, pow(self.h2nodes, -0.5), (self.h2nodes, self.h1nodes))

        # Hidden Layer 2 -> Output Layer weights
        self.wh2o  = np.random.normal(0.0, pow(self.onodes, -0.5), (self.onodes, self.h2nodes))

        # Sigmoid activation function
        self.activation_function = lambda x: scipy.special.expit(x)

        # ==============================================================
        # def activation(x):
        #      return scipy.special.expit(x)
        #
        #  self.activation_function = activation
        # ==============================================================

    def train(self, inputs_list, targets_list):
        # Convert input list into a 2D column vector
        inputs  = np.array(inputs_list,  ndmin=2).T
        targets = np.array(targets_list, ndmin=2).T

        # Forward Pass: Input -> Hidden 1 -> Hidden 2 -> Output
        h1_inputs  = np.dot(self.wih1, inputs)
        h1_outputs = self.activation_function(h1_inputs)

        h2_inputs  = np.dot(self.wh1h2, h1_outputs)
        h2_outputs = self.activation_function(h2_inputs)

        final_inputs  = np.dot(self.wh2o, h2_outputs)
        final_outputs = self.activation_function(final_inputs)

        # Backpropagation: send the error backwards
        output_errors = targets - final_outputs
        h2_errors     = np.dot(self.wh2o.T,  output_errors)
        h1_errors     = np.dot(self.wh1h2.T, h2_errors)

        # Update Hidden2 -> Output weights
        self.wh2o  += self.lr * np.dot((output_errors * final_outputs * (1.0 - final_outputs)),
                                       h2_outputs.T)

        # Update Hidden1 -> Hidden2 weights
        self.wh1h2 += self.lr * np.dot((h2_errors * h2_outputs * (1.0 - h2_outputs)),
                                       h1_outputs.T)

        # Update Input -> Hidden1 weights
        self.wih1  += self.lr * np.dot((h1_errors * h1_outputs * (1.0 - h1_outputs)),
                                       inputs.T)

    def query(self, inputs_list):
        inputs = np.array(inputs_list, ndmin=2).T

        h1_inputs  = np.dot(self.wih1, inputs)
        h1_outputs = self.activation_function(h1_inputs)

        h2_inputs  = np.dot(self.wh1h2, h1_outputs)
        h2_outputs = self.activation_function(h2_inputs)

        final_inputs  = np.dot(self.wh2o, h2_outputs)
        final_outputs = self.activation_function(final_inputs)

        return final_outputs
```

<tr>
<td style="background:#0B1E3D;color:#FFFFFF;border-radius:10px;padding:16px;width:33%;text-align:center"><code style="color:#00D9E9;font-size:18px">__init__</code><br><span style="font-size:15px;color:#BcD6E6">build the weights</span></td>
<td style="background:#0B1E3D;color:#FFFFFF;border-radius:10px;padding:16px;width:33%;text-align:center"><code style="color:#00D9E9;font-size:18px">train()</code><br><span style="font-size:15px;color:#BcD6E6">setter &mdash; learn, fix weights</span></td>
<td style="background:#0B1E3D;color:#FFFFFF;border-radius:10px;padding:16px;width:33%;text-align:center"><code style="color:#00D9E9;font-size:18px">query()</code><br><span style="font-size:15px;color:#BcD6E6">getter &mdash; give the answer</span></td>
</tr></table>

<div style="background:#FFFFFF;border:2px dashed #0B1E3D;border-radius:12px;padding:16px 24px;font-family:Segoe UI,system-ui,sans-serif;font-size:22px;color:#0B1E3D;font-weight:700;text-align:center">
  Same pattern as the very first class: &nbsp;<span style="color:#00E08A">BUILD</span> &rarr; <span style="color:#0a3d62">SHOW</span> &rarr; <span style="color:#FFB000">UPDATE</span>
</div>

<div style="background:#0a3d62;border-radius:10px;padding:14px 24px;font-family:Segoe UI,system-ui,sans-serif;margin-top:10px">
  <span style="color:#FFB000;font-size:30px;font-weight:900">11</span>
  <span style="color:#FFFFFF;font-size:27px;font-weight:800;margin-left:12px">Build the object &mdash; network configuration</span>
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">

Just like <code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">nike = Bag(...)</code>, we now make one real object from the blueprint.
<b>784</b> inputs because 28&times;28 = 784 pixels, and <b>10</b> outputs because there are ten digits.

</div>

```python
# STEP 1 — choose the size of the network and the learning rate
input_nodes   = 784      # 28 x 28 pixels
hidden1_nodes = 500
hidden2_nodes = 300
output_nodes  = 10       # digits 0 to 9
learning_rate = 0.15

# STEP 2 — build the object
n = NeuralNetwork(input_nodes, hidden1_nodes, hidden2_nodes,
                  output_nodes, learning_rate)

print(n)   # the network object is ready
```

<div style="background:#0a3d62;border-radius:10px;padding:14px 24px;font-family:Segoe UI,system-ui,sans-serif;margin-top:10px">
  <span style="color:#FFB000;font-size:30px;font-weight:900">12</span>
  <span style="color:#FFFFFF;font-size:27px;font-weight:800;margin-left:12px">Train the network</span>
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">

Two small preparations before every image is fed in:<br>
<b>1. Scale the inputs</b> &mdash; pixels 0&ndash;255 become 0.01&ndash;1.00, because a sigmoid cannot handle huge numbers.<br>
<b>2. Build the target</b> &mdash; ten values of 0.01, and only the correct digit becomes 0.99.

</div>

<div style="background:#FFFFFF;border-left:8px solid #00D9E9;border-radius:8px;padding:14px 22px;font-family:Segoe UI,system-ui,sans-serif;font-size:21px;color:#0B1E3D;font-weight:600">
  📌 One <b>epoch</b> = the network has seen all 20,000 images once. More epochs = more practice.
</div>

```python
# STEP 1 — load the training records into a list
with open("mnist_train_20000.csv", "r") as f:
    training_data_list = f.readlines()

print('training records :', len(training_data_list))
```

```python
# STEP 2 — the training loop
# Number of complete passes through the training dataset
epochs = 100          # for a quick live demo a smaller number works, e.g. 5

for e in range(epochs):

    for record in training_data_list:

        all_values = record.split(",")

        # scale the 784 pixels from 0-255 into 0.01-1.00
        inputs = (np.asarray(all_values[1:], dtype=float) / 255.0 * 0.99) + 0.01

        # target: 0.01 everywhere, 0.99 at the correct digit
        targets = np.zeros(output_nodes) + 0.01
        targets[int(all_values[0])] = 0.99

        n.train(inputs, targets)

    print(f"Epoch {e+1}/{epochs} Completed")
```

<div style="background:#0a3d62;border-radius:10px;padding:14px 24px;font-family:Segoe UI,system-ui,sans-serif;margin-top:10px">
  <span style="color:#FFB000;font-size:30px;font-weight:900">13</span>
  <span style="color:#FFFFFF;font-size:27px;font-weight:800;margin-left:12px">Test it &mdash; and score it</span>
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">

For every test image we call <code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">query()</code>, get ten numbers back, and take the
<b>biggest</b> one with <code style="background:#e6f0f6;color:#0a3d62;padding:2px 7px;border-radius:4px">np.argmax</code> &mdash; that index is the network's answer.

</div>

```python
# STEP 1 — load the unseen testing records
with open("mnist_test_3000.csv", "r") as f:
    test_data_list = f.readlines()

print('testing records :', len(test_data_list))
```

```python
# STEP 2 — ask the network about every test image
# Store 1 for correct prediction and 0 for incorrect prediction
scorecard = []

for record in test_data_list:

    all_values = record.split(",")

    correct_label = int(all_values[0])

    inputs = (np.asarray(all_values[1:], dtype=float) / 255.0 * 0.99) + 0.01

    outputs = n.query(inputs)

    predicted_label = np.argmax(outputs)

    scorecard.append(1 if predicted_label == correct_label else 0)
```

```python
# STEP 3 — the score
scorecard_array = np.asarray(scorecard)

# Calculate overall classification accuracy
accuracy = scorecard_array.sum() / scorecard_array.size

print(f"Correct Predictions : {scorecard_array.sum()} / {scorecard_array.size}")
print(f"Accuracy            : {accuracy * 100:.2f}%")
```

  💡 Accuracy is measured only on images the network has <b>never seen</b>. That is the real test of learning.
</div>

  <span style="color:#FFFFFF;font-size:27px;font-weight:800;margin-left:12px">See it with your own eyes</span>
</div>

<div style="font-family:Segoe UI,system-ui,sans-serif;font-size:16px;color:#33475a;line-height:1.7">

We reshape the 784 numbers back into a 28&times;28 picture and print the true label next to the prediction.
<b><span style="color:#0f8a68">Green</span></b> = correct, <b><span style="color:#c0392b">red</span></b> = wrong.

</div>

```python
# Display a few test images with predicted labels
num_examples = 6

fig, axes = plt.subplots(1, num_examples, figsize=(15, 3))

for i in range(num_examples):

    all_values = test_data_list[i].split(",")

    correct_label = int(all_values[0])

    image_array = np.asarray(all_values[1:], dtype=float).reshape(28, 28)

    inputs = (np.asarray(all_values[1:], dtype=float) / 255.0 * 0.99) + 0.01

    outputs = n.query(inputs)

    predicted_label = np.argmax(outputs)

    color = "green" if predicted_label == correct_label else "red"

    axes[i].imshow(image_array, cmap="gray")
    axes[i].set_title(f"True: {correct_label}\nPred: {predicted_label}", color=color)
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```
